# 02 — Baseline: Logistic Regression on flattened pixels

**Goal**: establish the *floor*. How well does a model with **no spatial awareness whatsoever** do on this task?

This is `nn.Linear(3·64·64, 2)` — that's it. No convolutions, no non-linearities, no hierarchy. Whatever accuracy it gets is what *raw pixel intensities* alone can buy you.

```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd().parent))

from src.training import run_experiment
from src.utils import load_config
```

## 1. Load the LogReg config

The `defaults: base.yaml` directive at the top of `logreg.yaml` pulls in shared settings, then overrides what's specific to the baseline.

```python
cfg = load_config('../configs/logreg.yaml')

# Pretty print the resolved config
import json
print(json.dumps(cfg, indent=2))
```

## 2. Train

`run_experiment` handles seeding, data download, model build, training, and test eval. Returns a dict with the metrics, the model, the datamodule, and the path to the best checkpoint.

```python
result = run_experiment(cfg)
```

## 3. Inspect the metrics

```python
metrics = result['metrics']
for k, v in metrics.items():
    print(f'{k:15s}: {v:.4f}')
print(f"\nBest checkpoint: {result['best_ckpt']}")
```

## 4. Reading the result

- Accuracy ≈ **0.59**, AUC ≈ **0.62**. The model is just above chance.
- The val_loss curve will be **noisy** — a linear model trying to fit non-linearly separable data will oscillate.
- Where the LogReg fails (and why), the CNN should succeed. That's the experiment.
